```python
from pyspark.sql import functions as F

customers_path = "/Volumes/retail_catalog/bronze/raw_files/customers.csv"
products_path = "/Volumes/retail_catalog/bronze/raw_files/products.csv"
orders_path = "/Volumes/retail_catalog/bronze/raw_files/orders.csv"

customers_df = spark.read.csv(customers_path, header=True, inferSchema=True)
product_df = spark.read.csv(products_path, header=True, inferSchema=True)
orders_df = spark.read.csv(orders_path, header=True, inferSchema=True)
display(customers_df)
display(product_df)
display(orders_df)
```

```python
customer_bronze = (customers_df
                   .withColumn('ingestion_timestamp',F.current_timestamp())
                   .withColumn('source_file',F.col('_metadata.file_path'))
                   .withColumn('batch_id',F.date_format(F.current_timestamp(),'yyyyMMddHHmmss'))
                )
product_bronze = (product_df
                  .withColumn('ingestion_timestamp',F.current_timestamp())
                  .withColumn('source_file',F.col('_metadata.file_path'))
                  .withColumn('batch_id',F.date_format(F.current_timestamp(),'yyyyMMddHHmmss'))
                  )
orders_bronze = (orders_df
                 .withColumn('ingestion_timestamp',F.current_timestamp())
                 .withColumn('source_file',F.col('_metadata.file_path'))
                 .withColumn('batch_id',F.date_format(F.current_timestamp(),'yyyyMMddHHmmss'))
                  )
```

```python
customer_bronze.write.format('delta').mode('overwrite').saveAsTable('retail_catalog.bronze.customers')
product_bronze.write.format('delta').mode('overwrite').saveAsTable('retail_catalog.bronze.products')
orders_bronze.write.format('delta').mode('overwrite').saveAsTable('retail_catalog.bronze.orders')
```
